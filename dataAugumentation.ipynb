```python
# =========================================================
# PROGRAM 1: MNIST CLASSIFICATION WITHOUT DATASET AUGMENTATION
# =========================================================

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense

# =========================================================
# STEP 1: Load MNIST dataset
# =========================================================

(x_train, y_train), (x_test, y_test) = \
    tf.keras.datasets.mnist.load_data()

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)

print("Test images:", x_test.shape)
print("Test labels:", y_test.shape)

# =========================================================
# STEP 2: Normalize pixel values
# =========================================================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# =========================================================
# STEP 3: Create Deep Neural Network
# =========================================================

model = Sequential([

    # Input layer
    Input(shape=(28, 28)),

    # Convert 28 x 28 image into 784 values
    Flatten(),

    # Hidden Layer 1
    Dense(128, activation="relu"),

    # Hidden Layer 2
    Dense(64, activation="relu"),

    # Output Layer
    Dense(10, activation="softmax")
])

# =========================================================
# STEP 4: Display model architecture
# =========================================================

model.summary()

# =========================================================
# STEP 5: Compile the model
# =========================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# =========================================================
# STEP 6: Train the model
# =========================================================

history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1
)

# =========================================================
# STEP 7: Evaluate on TEST dataset
# =========================================================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

# =========================================================
# STEP 8: Display result
# =========================================================

print("\n====================================")
print("RESULT WITHOUT AUGMENTATION")
print("====================================")

print("Test Loss     :", test_loss)
print("Test Accuracy :", test_accuracy)
```

```python
# =========================================================
# PROGRAM 2: MNIST CLASSIFICATION WITH DATASET AUGMENTATION
# =========================================================

import tensorflow as tf

from tensorflow.keras import Sequential

from tensorflow.keras.layers import (
    Input,
    Flatten,
    Dense,
    RandomRotation,
    RandomTranslation,
    RandomZoom
)

# =========================================================
# STEP 1: Load MNIST dataset
# =========================================================

(x_train, y_train), (x_test, y_test) = \
    tf.keras.datasets.mnist.load_data()

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)

print("Test images:", x_test.shape)
print("Test labels:", y_test.shape)

# =========================================================
# STEP 2: Normalize pixel values
# =========================================================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# =========================================================
# STEP 3: Create DATA AUGMENTATION
# =========================================================

data_augmentation = Sequential([

    # Slightly rotate the image
    RandomRotation(0.1),

    # Slightly move the image
    RandomTranslation(
        height_factor=0.1,
        width_factor=0.1
    ),

    # Slightly zoom the image
    RandomZoom(0.1)
])

# =========================================================
# STEP 4: Create Deep Neural Network
# =========================================================

model = Sequential([

    # Input layer
    Input(shape=(28, 28)),

    # Data augmentation
    data_augmentation,

    # Convert 28 x 28 into 784 values
    Flatten(),

    # Hidden Layer 1
    Dense(128, activation="relu"),

    # Hidden Layer 2
    Dense(64, activation="relu"),

    # Output Layer
    Dense(10, activation="softmax")
])

# =========================================================
# STEP 5: Display model architecture
# =========================================================

model.summary()

# =========================================================
# STEP 6: Compile the model
# =========================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# =========================================================
# STEP 7: Train the model
# =========================================================

history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1
)

# =========================================================
# STEP 8: Evaluate on TEST dataset
# =========================================================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

# =========================================================
# STEP 9: Display result
# =========================================================

print("\n====================================")
print("RESULT WITH AUGMENTATION")
print("====================================")

print("Test Loss     :", test_loss)
print("Test Accuracy :", test_accuracy)
```

```python
# ==========================================================
# PROGRAM 3: MNIST DATASET AUGMENTATION USING A DEEP
#            NEURAL NETWORK
# ==========================================================

import tensorflow as tf
import numpy as np

from tensorflow.keras import Sequential

from tensorflow.keras.layers import (
    Input,
    Flatten,
    Dense
)

# ==========================================================
# STEP 1: LOAD MNIST DATASET
# ==========================================================

(x_train, y_train), (x_test, y_test) = \
    tf.keras.datasets.mnist.load_data()

print("Original training dataset:", x_train.shape)
print("Original test dataset:", x_test.shape)

# ==========================================================
# STEP 2: SELECT 10,000 ORIGINAL TRAINING IMAGES
# ==========================================================

N = 10000

x_original = x_train[:N]
y_original = y_train[:N]

print("\nSelected original images:", x_original.shape)
print("Selected original labels:", y_original.shape)

# ==========================================================
# STEP 3: CREATE ROTATED IMAGES
# ==========================================================

# Convert images to floating point
x_float = x_original.astype("float32")

# Add channel dimension
x_float = x_float[..., np.newaxis]

# Rotate each image by 10 degrees
x_rotated = tf.keras.layers.RandomRotation(
    factor=10 / 360,
    fill_mode="constant"
)(x_float, training=True)

# Convert TensorFlow tensor to NumPy array
x_rotated = x_rotated.numpy()

# Remove channel dimension
x_rotated = np.squeeze(x_rotated, axis=-1)

# Convert back to uint8
x_rotated = np.clip(
    x_rotated,
    0,
    255
).astype("uint8")

# Labels remain unchanged
y_rotated = y_original.copy()

print("\nRotated images:", x_rotated.shape)
print("Rotated labels:", y_rotated.shape)

# ==========================================================
# STEP 4: CREATE SHIFTED IMAGES
# ==========================================================

# Shift images horizontally and vertically
x_shifted = tf.keras.layers.RandomTranslation(
    height_factor=0.10,
    width_factor=0.10,
    fill_mode="constant"
)(x_float, training=True)

# Convert TensorFlow tensor to NumPy
x_shifted = x_shifted.numpy()

# Remove channel dimension
x_shifted = np.squeeze(x_shifted, axis=-1)

# Keep pixel values between 0 and 255
x_shifted = np.clip(
    x_shifted,
    0,
    255
).astype("uint8")

# Labels remain unchanged
y_shifted = y_original.copy()

print("Shifted images:", x_shifted.shape)
print("Shifted labels:", y_shifted.shape)

# ==========================================================
# STEP 5: COMBINE ORIGINAL + ROTATED + SHIFTED DATA
# ==========================================================

x_augmented = np.concatenate(
    [
        x_original,
        x_rotated,
        x_shifted
    ],
    axis=0
)

y_augmented = np.concatenate(
    [
        y_original,
        y_rotated,
        y_shifted
    ],
    axis=0
)

# ==========================================================
# STEP 6: DISPLAY DATASET SIZE
# ==========================================================

print("\n==========================================")
print("DATASET SIZE AFTER AUGMENTATION")
print("==========================================")

print("Original images :", len(x_original))
print("Rotated images  :", len(x_rotated))
print("Shifted images  :", len(x_shifted))

print("------------------------------------------")

print("Final images    :", len(x_augmented))
print("Final labels    :", len(y_augmented))

# ==========================================================
# STEP 7: NORMALIZE THE AUGMENTED DATA
# ==========================================================

x_augmented = x_augmented.astype("float32") / 255.0

x_test = x_test.astype("float32") / 255.0

# ==========================================================
# STEP 8: CREATE DEEP NEURAL NETWORK
# ==========================================================

model = Sequential([

    # Input Layer
    Input(shape=(28, 28)),

    # Convert 28 x 28 into 784
    Flatten(),

    # Hidden Layer 1
    Dense(
        128,
        activation="relu"
    ),

    # Hidden Layer 2
    Dense(
        64,
        activation="relu"
    ),

    # Output Layer
    Dense(
        10,
        activation="softmax"
    )
])

# ==========================================================
# STEP 9: DISPLAY ARCHITECTURE
# ==========================================================

print("\n==========================================")
print("DEEP NEURAL NETWORK")
print("==========================================")

model.summary()

# ==========================================================
# STEP 10: COMPILE THE MODEL
# ==========================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# ==========================================================
# STEP 11: TRAIN THE MODEL
# ==========================================================

history = model.fit(
    x_augmented,
    y_augmented,
    epochs=5,
    batch_size=128,
    validation_split=0.1
)

# ==========================================================
# STEP 12: TEST THE MODEL
# ==========================================================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

# ==========================================================
# STEP 13: DISPLAY FINAL RESULT
# ==========================================================

print("\n==========================================")
print("FINAL RESULT")
print("==========================================")

print("Test Loss     :", test_loss)
print("Test Accuracy :", test_accuracy)
```
